```sql
SELECT c.name AS company, r.canonical_tag, r.ddate, r.qtrs,
       r.prev_value, r.revised_value,
       round(r.pct_change * 100, 1) AS pct_change,
       r.revised_at_ts, r.source_form
FROM hindsight_dev.gold.fact_revision r
LEFT JOIN hindsight_dev.gold.dim_company_scd2 c
       ON r.cik = c.cik AND c.is_current
WHERE abs(r.prev_value) > 1e8
ORDER BY abs(r.pct_change) DESC
LIMIT 200;
```
